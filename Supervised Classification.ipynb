**Q-1 What is Information Gain, and how is it used in Decision Trees?**

Ans - Information Gain is a measure of how much information about the class label is gained by splitting a dataset based on a particular attribute. It is calculated as the difference between the entropy of the original dataset and the weighted entropy after splitting the dataset on an attribute.

Entropy measures the amount of randomness or impurity in the data. A dataset with mixed class labels has high entropy, while a dataset with only one class has zero entropy. Information Gain quantifies how effectively an attribute reduces this entropy.

**Use of Information Gain in Decision Trees**

Information Gain is a key concept used in Decision Tree algorithms in Machine Learning to select the best attribute for splitting data. Decision Trees work by repeatedly dividing a dataset into smaller subsets so that the data in each subset becomes more homogeneous. Information Gain helps in choosing the attribute that results in the maximum reduction in uncertainty after the split.

In a Decision Tree, Information Gain is used at each node to decide which attribute should be selected for splitting the data. The algorithm calculates the Information Gain for all available attributes and chooses the attribute with the highest Information Gain. This ensures that each split makes the dataset purer and improves classification accuracy. The process is repeated recursively until a stopping condition is met.

**Q-2  What is the difference between Gini Impurity and Entropy?**

**Hint: Directly compares the two main impurity measures, highlighting strengths, weaknesses, and appropriate use cases.**

Ans - Gini Impurity and Entropy are two important impurity measures used in Decision Tree algorithms to evaluate the quality of a split. Both aim to measure how mixed the class labels are at a node, but they differ in their interpretation, calculation, and practical usage.

**Difference Between Gini Impurity and Entropy**

**Gini Impurity** measures the probability that a randomly selected data point from a node would be incorrectly classified if it were labeled according to the class distribution at that node. It is calculated as one minus the sum of the squared probabilities of each class. A Gini value of zero indicates a completely pure node, while higher values indicate greater impurity.

* Gini Impurity is computationally efficient because it does not involve logarithmic calculations, making it faster for large datasets. It is mainly used in the CART (Classification and Regression Tree) algorithm.

**Entropy**, on the other hand, measures the amount of uncertainty or randomness in a dataset. It quantifies how unpredictable the class labels are at a node and is calculated using the negative sum of the probabilities of each class multiplied by the logarithm of those probabilities. Like Gini Impurity, entropy is zero for a pure node and increases as the classes become more evenly distributed.

* Entropy is more sensitive to changes in class probabilities and provides a more informative measure of impurity, but it is computationally more expensive due to the use of logarithmic functions. Entropy is commonly used in ID3 and C4.5 decision tree algorithms.

**Q-3 What is Pre-Pruning in Decision Trees?**

Ans - Pre-pruning, also known as early stopping, is the process of halting further splitting of a node while building the decision tree if certain predefined conditions are met. Instead of allowing the tree to grow until all nodes become pure, pre-pruning restricts the depth or size of the tree during training itself.

**Why Pre-Pruning is Used**

Pre-pruning is used to reduce overfitting, improve generalization, and make the model simpler and faster. By avoiding unnecessary splits, the tree becomes easier to interpret and performs better on unseen data.

**Advantages**

* Reduces overfitting

* Improves model generalization

* Faster training time

* Produces simpler and more interpretable trees

**Q-4 Write a Python program to train a Decision Tree Classifier using Gini Impurity as the criterion and print the feature importances (practical). Hint: Use criterion='gini' in DecisionTreeClassifier and access .feature_importances_.**

```python
# Import required libraries
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier

# Load dataset
data = load_iris()
X = data.data          # Features
y = data.target        # Target labels
feature_names = data.feature_names

# Create Decision Tree Classifier using Gini Impurity
dt = DecisionTreeClassifier(criterion='gini', random_state=42)

# Train the model
dt.fit(X, y)

# Print feature importances
print("Feature Importances:")
for name, importance in zip(feature_names, dt.feature_importances_):
    print(f"{name}: {importance:.4f}")
```

**Q-5 What is a Support Vector Machine (SVM)?**

Ans - A Support Vector Machine (SVM) is a supervised machine learning algorithm used for classification and regression tasks. It is mainly known for its strong theoretical foundation and its ability to work well with high-dimensional data.

An SVM works by finding an optimal decision boundary, called a hyperplane, that separates data points of different classes. The goal of SVM is to choose the hyperplane that maximizes the margin, which is the distance between the hyperplane and the nearest data points from each class.

**Advantages**

* Effective in high-dimensional spaces

* Works well with small and medium-sized datasets

* Strong generalization capability

**Limitation**

* Computationally expensive for very large datasets

* Requires careful selection of kernel and parameters

**Q-6 What is the Kernel Trick in SVM?**

Ans - The Kernel Trick is a powerful technique used in Support Vector Machines (SVMs) to handle non-linearly separable data. It allows SVMs to create non-linear decision boundaries without explicitly transforming the data into a higher-dimensional space.

The Kernel Trick works by implicitly mapping the input data from its original feature space into a higher-dimensional feature space where the data becomes linearly separable. Instead of computing this transformation directly, SVM uses a kernel function that computes the inner product between data points in the transformed space.

This makes the computation efficient and avoids the high cost of explicit feature transformation.

**Why the Kernel Trick is Needed**

In many real-world problems, data cannot be separated by a straight line or flat plane. A simple linear SVM fails in such cases. The Kernel Trick enables SVM to:

* Learn complex, non-linear decision boundaries

* Handle curved or irregular data patterns

* Maintain computational efficiency

Q-7  Write a Python program to train two SVM classifiers with Linear and RBF kernels on the Wine dataset, then compare their accuracies.

Hint:Use SVC(kernel='linear') and SVC(kernel='rbf'), then compare accuracy scores after fitting on the same dataset.

Ans -

```python
# Import required libraries
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# Load the Wine dataset
data = load_wine()
X = data.data      # Features
y = data.target    # Labels

# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Train SVM with Linear kernel
svm_linear = SVC(kernel='linear')
svm_linear.fit(X_train, y_train)
y_pred_linear = svm_linear.predict(X_test)

# Train SVM with RBF kernel
svm_rbf = SVC(kernel='rbf')
svm_rbf.fit(X_train, y_train)
y_pred_rbf = svm_rbf.predict(X_test)

# Calculate accuracies
linear_accuracy = accuracy_score(y_test, y_pred_linear)
rbf_accuracy = accuracy_score(y_test, y_pred_rbf)

# Print results
print("Accuracy with Linear Kernel SVM:", linear_accuracy)
print("Accuracy with RBF Kernel SVM:", rbf_accuracy)
```

**Q-8 What is the Naïve Bayes classifier, and why is it called "Naïve"?**

Ans - The Naïve Bayes classifier is a probabilistic supervised machine learning algorithm mainly used for classification tasks. It is based on Bayes’ Theorem and is especially effective for high-dimensional data such as text classification and spam detection.

The classifier is called “Naïve” because it makes a strong and simplifying assumption that all input features are conditionally independent of each other given the class label.

In reality, features are often correlated, but Naïve Bayes ignores these dependencies and treats each feature as contributing independently to the final prediction.

**Advantages**

* Simple and fast

* Works well with large datasets

* Performs well even with limited training data

**Limitation**

* Poor performance when features are highly correlated

**Q-9 Explain the differences between Gaussian Naïve Bayes, Multinomial Naïve Bayes, and Bernoulli Naïve Bayes**

Ans - Naïve Bayes classifiers are a family of probabilistic algorithms based on Bayes’ Theorem. The main difference between Gaussian, Multinomial, and Bernoulli Naïve Bayes lies in the assumption they make about the distribution of feature values. Each variant is suitable for different types of data.

**Gaussian Naïve Bayes**

Gaussian Naïve Bayes assumes that the continuous features of the dataset follow a normal (Gaussian) distribution within each class. For every feature and class, the algorithm estimates the mean and variance, which are then used to compute the probability of a data point belonging to a class.

This variant is best suited for continuous numerical data, such as measurements, sensor readings, or real-valued features. It is commonly used in problems where features can take any real value.

**Multinomial Naïve Bayes**

Multinomial Naïve Bayes assumes that the features represent counts or frequencies. It models how often a feature appears in a sample and is especially popular in text classification tasks, where features correspond to word counts or term frequencies.

This classifier works well with discrete, non-negative integer features and is widely used for applications like spam detection, document classification, and topic modeling.

**Bernoulli Naïve Bayes**

Bernoulli Naïve Bayes assumes that features are binary, meaning they can take only two values, such as 0 or 1. It models the presence or absence of a feature rather than its frequency.

This variant is suitable for data where the existence of a feature is more important than how often it appears, such as whether a word occurs in a document or not.

**Key Differences in Usage**

Gaussian Naïve Bayes is used for continuous-valued data, Multinomial Naïve Bayes is used for count-based or frequency data, and Bernoulli Naïve Bayes is used for binary data. Choosing the correct variant depends on the nature of the input features and the problem being solved.

**Q-10 Write a Python program to train a Gaussian Naïve Bayes classifier on the Breast Cancer dataset and evaluate accuracy.**

**Hint:Use GaussianNB() from sklearn.naive_bayes and the Breast Cancer dataset from sklearn.datasets.**

```python
# Import required libraries
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score

# Load the Breast Cancer dataset
data = load_breast_cancer()
X = data.data      # Features
y = data.target    # Target labels

# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Create Gaussian Naïve Bayes classifier
gnb = GaussianNB()

# Train the model
gnb.fit(X_train, y_train)

# Make predictions
y_pred = gnb.predict(X_test)

# Evaluate accuracy
accuracy = accuracy_score(y_test, y_pred)

# Print the accuracy
print("Accuracy of Gaussian Naïve Bayes Classifier:", accuracy)
```
