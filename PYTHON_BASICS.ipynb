Q-2 What is an interpreter in Python?
  - An interpreter in Python is a program that reads and executes Python code line by line. Unlike compiled languages that translate the entire program at once, Python uses an interpreter to run the code directly.

- When you write and run a Python program, the interpreter translates your code into machine language step by step. This makes it easier to find and fix errors quickly, which is helpful for beginners and developers.

Q-3 What are pre-defined keywords in Python?
- Pre-defined keywords in Python are reserved words that have special meaning to the Python interpreter. These words are used to define the structure and rules of the Python language, and you cannot use them as variable names, function names, or identifiers.

- Each keyword performs a specific function, such as creating loops, making decisions, or defining data types.For ex - if, for, while.

Q-4 Can keywords be used as variable names?
- No, keywords cannot be used as variable names in Python. Keywords are
reserved words that have special meaning in the language, like if, for, while, and def. Because Python uses them to understand the structure and commands in your code, using them as variable names would confuse the interpreter and cause errors.

Q-5 What is mutability in Python?

Mutability in Python means whether or not an object’s value can be changed after it is created.

- If an object can be changed, it is called mutable.

- If an object cannot be changed, it is called immutable.

For example, lists and dictionaries are mutable because you can add, remove, or change their items. But strings and tuples are immutable, meaning once you create them, you cannot change their content.

Q-6 Why are lists mutable, but tuples are immutable?

Lists and tuples are both used to store collections of items in Python, but they behave differently because of how they are designed.

- Lists are mutable because they are made to be flexible. You can add, remove, or change items in a list whenever you want. This makes lists useful when you need to update or modify data.

- Tuples are immutable because they are meant to be fixed and unchangeable. Once you create a tuple, you cannot change its contents. This makes tuples safer to use when you want to make sure the data stays the same throughout the program.

Q-7 What is the difference between “==” and “is” operators in Python?

- == - checks if the values of two objects are equal. It compares the content inside the objects.

- is - checks if two variables point to the same object in memory. It tests whether they are actually the exact same thing.

Q-8 What are logical operators in Python?

Logical operators are special symbols or words in Python that help you combine multiple conditions or make decisions based on true or false values. They work with Boolean values (True or False).

Python has three main logical operators:

- and – Returns True if both conditions are true

- or – Returns True if at least one condition is true

- not – Reverses the result; returns True if the condition is false, and vice versa

Q-9 What is type casting in Python?

- Type casting means converting data from one type to another, like changing a number stored as a string "123" into an integer 123. This helps when you need to do operations that require specific data types.

Q-10 What is the difference between implicit and explicit type casting?

- Implicit type casting happens automatically when Python converts one data type to another to avoid errors. For example, when adding an integer and a float, Python changes the integer to float automatically.

- Explicit type casting is when the programmer manually converts a value using functions like int(), float(), or str(). For example, changing a string "10" to an integer using int("10").

  For example, you might want your program to check a user’s age and decide if they can vote. If the age is 18 or more, the program shows a message saying they can vote. If it’s less than 18, it tells them they are too young.

Q-12 How does the elif statement work?

- elif stands for “else if” and is used after an if to check additional conditions if the first one is false. Python will check each elif condition in order and run the code for the first true one.

Q-13 What is the difference between for and while loops?

- for loops run a set number of times, often looping over items in a list or numbers in a range.

- while loops keep running as long as a condition is true, which is useful when you don’t know beforehand how many times you need to repeat.

Q-14 Describe a scenario where a while loop is more suitable than a for loop?

- A while loop is more suitable than a for loop when you don’t know in advance how many times you need to repeat an action, and the repetition depends on a condition that changes during the program’s execution.

  Scenario Example:

  Imagine you are creating a program that asks a user to enter a password. You want the program to keep asking until the user enters the correct password. Since you don’t know how many attempts the user will take, a while loop works best.

  In this case, the loop continues running as long as the user hasn’t typed the right password. A for loop wouldn’t be suitable here because you don’t know how many tries the user will need ahead of time.

  - EXAMPLE BELOW


```python
correct_password = "python123"
user_input = ""

while user_input != correct_password:
    user_input = input("Enter the password: ")
    if user_input != correct_password:
        print("Wrong password, try again.")
print("Access granted!")
```

Q-5 Create a list and tuple in Python, and demonstrate how attempting to change an element works differently for each.

```python
my_list = [1, 2, 3]
my_tuple = (1, 2, 3)

# Change list element
my_list[0] = 10
print("Modified list:", my_list)

# Attempt to change tuple element (will cause error)
try:
    my_tuple[0] = 10
except TypeError as e:
    print("Error when modifying tuple:", e)
```

Q-6 Write a function to demonstrate the behavior of mutable and immutable arguments.

```python
def modify_immutable(x):
    x = x + 1
    print("Inside function (immutable):", x)

def modify_mutable(lst):
    lst.append(4)
    print("Inside function (mutable):", lst)

num = 10
modify_immutable(num)
print("Outside function:", num)

my_list = [1, 2, 3]
modify_mutable(my_list)
print("Outside function:", my_list)
```

Q-14  Write a program to reverse a string using a while loop.

```python
string = input("Enter a string: ")
reversed_str = ""
index = len(string) - 1

while index >= 0:
    reversed_str += string[index]
    index -= 1

print("Reversed string:", reversed_str)
```

Q-15 Write a Python program to calculate the factorial of a number provided by the user using a while loop.

```python
num = int(input("Enter a number: "))
factorial = 1
i = 1

while i <= num:
    factorial *= i
    i += 1

print(f"Factorial of {num} is {factorial}")
```
