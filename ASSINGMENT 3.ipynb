**Q-1**  What is the difference between a function and a method in Python?

Ans - **Function in Python**

A function is a block of reusable code that performs a specific task. It is defined using the def keyword and can take zero or more arguments. Functions are not associated with any object or class unless explicitly used within one.

**Characteristics of Functions:**

* Defined using def keyword.

* Can be called independently.

* Not tied to any object or class unless nested in one.

**Method in Python**

A method is a function that is defined inside a class and is associated with the objects of that class. It automatically takes the instance of the class (self) as its first parameter when called.

**Characteristics of Methods:**

* Defined inside a class.

* Always takes self as the first parameter (for instance methods).

* Called using object instances (e.g., object.method()).

**Q-2 Explain the concept of function arguments and parameters in Python.**

Ans-

**Parameters** -

Parameters are the variable names listed inside the parentheses in a function’s definition. They act as placeholders that represent the values a function will use when it runs. Parameters define what kind of input a function expects.

**Arguments** -

Arguments are the actual values that are supplied to a function when it is called. These values are assigned to the corresponding parameters so the function can perform operations using them.

**Difference Between Parameters and Arguments**

* Parameters appear in the function definition.

* Arguments are used in the function call.

* Parameters receive values from the arguments during execution.

**Types of Arguments in Python**

Python supports several types of arguments:

* Positional Arguments – Passed in the same order as the parameters are defined.

* Keyword Arguments – Specified by parameter name, allowing flexibility in order.

* Default Arguments – Have a predefined value if no argument is provided.

* Variable-Length Arguments – Allow functions to accept an arbitrary number of values, either positional or keyword-based.

**Q-5 What are iterators in Python and how do they differ from iterables?**

**Ans** - **Iterable**

* An iterable is any Python object capable of returning its elements one at a time, allowing it to be looped over in a for loop or used in functions like list(), tuple(), etc.

* Examples of iterables include lists, tuples, strings, dictionaries, and sets.

* An iterable implements the __iter__() method, which returns an iterator


**Difference Between Iterators and Iterables**

Iterable:
* An iterable is any Python object that can return an iterator. It can be looped over multiple times because it can create a new iterator each time. Examples include lists, tuples, and strings.

Iterator:
* An iterator is an object that represents a stream of data and returns one element at a time when asked. It keeps track of its current position and can only be used once; once all elements are exhausted, it cannot restart.



**Q-6  Explain the concept of generators in Python and how they are defined.**

**Ans** - Generators are special types of iterators that generate values on the fly, one at a time, only when requested. They help save memory because they don’t store the entire sequence in memory but produce each value lazily.

**Generators are defined using:**

**Generator functions**: Functions that use the yield keyword to produce a sequence of values over time, pausing between each value until the next one is requested.

**Generator expressions**: Similar to list comprehensions but with parentheses instead of brackets.

**Q-7 What are the advantages of using generators over regular functions?**

Ans -

* **Memory Efficient**: Generators produce items one at a time, so they don’t require storing the entire data in memory.

* **Lazy Evaluation**: Values are computed only when needed, which can improve performance for large data sets.

* **Represent Infinite** **Sequences**: Generators can model infinite sequences since they don’t compute all values upfront.

* **Simpler Code**: Using yield can make writing iterators more straightforward compared to implementing an iterator class.

**Q-8 What is a lambda function in Python and when is it typically used ?**

Ans - A lambda function is a small anonymous function defined using the lambda keyword. It can have any number of arguments but only one expression, which is returned automatically.

**Typical uses:**

* When a simple, short function is needed without formally defining it with def.

* Commonly used as arguments to higher-order functions like map(), filter(), and sorted().

**Q-9 Explain the purpose and usage of the map() function in Python.**

Ans - The map() function applies a given function to each item in an iterable (like a list) and returns an iterator with the results.

* **Purpose**: To transform all items in a collection by applying the same operation to each one, without writing an explicit loop.

* **Usage**: You provide a function and one or more iterables to map(). It processes each element and returns the mapped results lazily (as an iterator).

**Q-1 Write a Python function that takes a list of numbers as input and returns the sum of all even numbers in the list.**

```python
def sum_even_numbers(numbers):
    total = 0
    for num in numbers:
        if num % 2 == 0:
            total += num
    return total

# Taking input from user as a space-separated string of numbers
user_input = input("Enter numbers separated by spaces: ")

# Convert input string into a list of integers
numbers_list = list(map(int, user_input.split()))

# Call the function and print the result
result = sum_even_numbers(numbers_list)
print("Sum of even numbers:", result)
```

**Q-2 Create a Python function that accepts a string and returns the reverse of that string.**

```python
def reverse_string():
    s = input("Enter a string: ")
    return s[::-1]

print("Reversed string:", reverse_string())
```

**Q-3 Implement a Python function that takes a list of integers and returns a new list containing the squares of each number.**

```python
def square_list():
    user_input = input("Enter integers separated by spaces: ")
    numbers = list(map(int, user_input.split()))
    squared = [num ** 2 for num in numbers]
    return squared

print("Squares:", square_list())
```

**Q-4 Write a Python function that checks if a given number is prime or not from 1 to 200.**

```python
def is_prime():
    num = int(input("Enter a number between 1 and 200: "))
    if num < 1 or num > 200:
        return "Number out of range."
    if num <= 1:
        return f"{num} is not a prime number."
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return f"{num} is not a prime number."
    return f"{num} is a prime number."

print(is_prime())
```

**Q-5  Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms.**

```python
class FibonacciIterator:
    def __init__(self, n):
        self.n = n
        self.i = 0
        self.a = 0
        self.b = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.i >= self.n:
            raise StopIteration
        if self.i == 0:
            self.i += 1
            return 0
        elif self.i == 1:
            self.i += 1
            return 1
        else:
            self.a, self.b = self.b, self.a + self.b
            self.i += 1
            return self.a

# Example:
fib = FibonacciIterator(5)
for num in fib:
    print(num)

```

**Q-6 Write a generator function in Python that yields the powers of 2 up to a given exponent**

```python
def powers_of_two():
    n = int(input("Enter the exponent: "))
    for i in range(n + 1):
        yield 2 ** i

for power in powers_of_two():
    print(power)
```

**Q-7  Implement a generator function that reads a file line by line and yields each line as a string.**

```python
def read_file_line_by_line(filename):
    with open(filename, 'r') as file:
        for line in file:
            yield line.rstrip('\n')

# Example usage:
filename = input("Enter the filename: ")
for line in read_file_line_by_line(filename):
    print(line)
```

**Q-8 Use a lambda function in Python to sort a list of tuples based on the second element of each tuple.**

```python
user_input = input("Enter tuples separated by commas (e.g. (1,3),(4,1)): ")
tuples_list = [tuple(map(int, t.strip("() ").split(','))) for t in user_input.split('),')]
sorted_list = sorted(tuples_list, key=lambda x: x[1])
print("Sorted list:", sorted_list)
```

**Q-9 Write a Python program that uses `map()` to convert a list of temperatures from Celsius to Fahrenheit.**

```python
def celsius_to_fahrenheit(c):
    return (c * 9/5) + 32

user_input = input()
celsius_temps = list(map(float, user_input.split()))
fahrenheit_temps = list(map(celsius_to_fahrenheit, celsius_temps))

print("Temperatures in Fahrenheit:", fahrenheit_temps)
```

**Q-10 Create a Python program that uses `filter()` to remove all the vowels from a given string.**

```python
def is_not_vowel(char):
    return char.lower() not in 'aeiou'

user_input = input("Enter a string: ")
filtered_chars = filter(is_not_vowel, user_input)
result = ''.join(filtered_chars)

print("String without vowels:", result)
```
