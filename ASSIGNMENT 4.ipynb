**Q-8 How is encapsulation achieved in Python?**

**Q-10 What are class and static methods in Python?**

* Defined with @staticmethod decorator

* Behaves like a normal function but inside a class

* Doesn’t access self or cls

**Q-15 What is an abstract class in Python**

**Q-17 What is the difference between a class variable and an instance variable?**

Ans -
* Class Variable: Declared outside any method within the class. It is shared by all instances of the class and can be accessed using the class name. It stores data that is constant or common to all objects.

* Instance Variable: Declared inside a method (typically __init__) using the self prefix. It is unique to each individual instance (object) and stores the specific state or data for that object.

**Q-18 What is multiple inheritance in Python?**

Ans - Multiple inheritance is the feature where a class can inherit attributes and methods from more than one parent class. Python fully supports this. When a method is called, Python determines the correct search order using the Method Resolution Order (MRO), which is crucial for resolving potential naming conflicts across the parent classes.

**Q-19 Explain the purpose of ‘’__str__’ and ‘__repr__’ ‘ methods in Python?**

Ans - These are special methods for defining an object's string representation:

* __str__ : Intended for readability and the end-user. It's automatically called by the print() function and should return a concise, human-readable description of the object.

* __repr__ : Intended for unambiguity and the developer/debugger. It should, if possible, return a string that looks like a valid Python expression that could recreate the object. It's what you see when you just type the object's name in the interpreter.



Q-20 What is the significance of the ‘super()’ function in Python?

Ans - The super() function is used within a subclass to call a method from its parent (or "superclass"). Its significance is two-fold:

* It allows a subclass to extend or safely override a parent's method (like __init__) while still executing the parent's original functionality.

* Crucially, in complex hierarchies (especially with multiple inheritance), super() doesn't just point to the immediate parent; it correctly follows the MRO to ensure the next class in the inheritance path is called, promoting robust and cooperative inheritance.

**Q-21 What is the significance of the __del__ method in Python?**

Ans - The __del__ method is the destructor. It's automatically called by Python's garbage collector when an object is about to be finally destroyed (i.e., when its reference count drops to zero). Its primary significance is to define cleanup actions for external resources the object might be holding, such as closing open files, releasing network sockets, or disconnecting from a database.

**Q-22 What is the difference between @staticmethod and @classmethod in Python?**

Ans -
* @staticmethod: This method takes no special first argument. It behaves like a plain function that happens to be defined inside a class. It cannot access or modify instance data or class data. It’s used for utility functions that are logically related to the class but don't depend on it.

* @classmethod: This method takes the class itself as the first argument, conventionally named cls. Because it receives the class, it can access and modify class variables. It's often used for factory methods that create instances in a specific way.

**Q-23 How does polymorphism work in Python with inheritance?**

Ans - Polymorphism in Python with inheritance works through Method Overriding. When a subclass redefines a method that already exists in its parent class, it overrides it. Python uses dynamic dispatch; when that method is called on an object, the interpreter figures out which specific implementation to run based on the object's actual type (the instance's class) at runtime, allowing the same method call (animal.speak()) to execute different code for different object types (Dog vs. Cat).

**Q-24 What is method chaining in Python OOP?**

Ans - Method chaining is a technique that lets you call multiple methods sequentially on the same object in a single line (e.g., obj.setup().configure().run()). This is made possible when each method, after completing its operation, returns the object itself via the statement return self.

 This fluent interface pattern improves code readability and conciseness, especially in builder patterns or sequence operations.

**Q-25 What is the purpose of the __call__ method in Python?**

Ans - The __call__ method allows an instance of a class to be treated like a function. If a class defines this method, you can execute the object directly using parentheses, like my_object().

The purpose is to create callable objects (sometimes called functors) that can maintain internal state while still being usable in contexts where a simple function is expected, such as complex decorators or stateful processing engines.

**Q-1 Create a parent class Animal with a method speak() that prints a generic message. Create a child class Dog that overrides the speak() method to print "Bark!".**

```python
print("1. Inheritance: Animal and Dog")
class Animal:
    def speak(self):
        print("Animal speaks")

class Dog(Animal):
    def speak(self):
        print("Bark!")

a = Animal()
d = Dog()
a.speak()
d.speak()
print("-" * 40)
```

**Q-2 Write a program to create an abstract class Shape with a method area(). Derive classes Circle and Rectangle from it and implement the area() method in both.**

```python
print("2. Abstract Class: Shape, Circle, Rectangle")
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

radius = float(input("Enter radius of circle: "))
c = Circle(radius)
print("Circle area:", c.area())

width = float(input("Enter width of rectangle: "))
height = float(input("Enter height of rectangle: "))
r = Rectangle(width, height)
print("Rectangle area:", r.area())
print("-" * 40)
```

Q-3  Implement a multi-level inheritance scenario where a class Vehicle has an attribute type. Derive a class Car and further derive a class ElectricCar that adds a battery attribute.

```python
print("\n3. Multilevel Inheritance")
class Vehicle:
    def __init__(self, t): self.type = t
class Car(Vehicle):
    def __init__(self, t, b): super().__init__(t); self.brand = b
class ElectricCar(Car):
    def __init__(self, t, b, bat): super().__init__(t, b); self.battery = bat
ec = ElectricCar(input("Type: "), input("Brand: "), input("Battery: "))
print(f"Type: {ec.type}, Brand: {ec.brand}, Battery: {ec.battery}")
```

**Q-4  Demonstrate polymorphism by creating a base class Bird with a method fly(). Create two derived classes Sparrow and Penguin that override the fly() method.**

```python
print("4. Polymorphism - Bird, Sparrow, Penguin")

class Bird:
    def fly(self):
        print("Some birds can fly")

class Sparrow(Bird):
    def fly(self):
        print("Sparrow flies high")

class Penguin(Bird):
    def fly(self):
        print("Penguins can't fly")

bird_type = input("Enter bird type (sparrow/penguin): ").strip().lower()

if bird_type == "sparrow":
    b = Sparrow()
elif bird_type == "penguin":
    b = Penguin()
else:
    b = Bird()

b.fly()
```

**Q-5 Write a program to demonstrate encapsulation by creating a class BankAccount with private attributes balance and methods to deposit, withdraw, and check balance.**

```python
print("\n5. Encapsulation")
class BankAccount:
    def __init__(self): self.__balance = 0
    def deposit(self, amt): self.__balance += amt
    def withdraw(self, amt):
        if amt <= self.__balance: self.__balance -= amt
        else: print("Insufficient funds")
    def get_balance(self): return self.__balance
acc = BankAccount()
acc.deposit(float(input("Deposit: ")))
acc.withdraw(float(input("Withdraw: ")))
print("Balance:", acc.get_balance())
```

**Q-6 Demonstrate runtime polymorphism using a method play() in a base class Instrument. Derive classes Guitar and Piano that implement their own version of play().**

```python
print("6. Runtime Polymorphism - Instrument, Guitar, Piano")

class Instrument:
    def play(self):
        print("Instrument is playing")

class Guitar(Instrument):
    def play(self):
        print("Guitar is playing a melody")

class Piano(Instrument):
    def play(self):
        print("Piano is playing a tune")

instrument_name = input("Choose an instrument (guitar/piano): ").strip().lower()

if instrument_name == "guitar":
    instr = Guitar()
elif instrument_name == "piano":
    instr = Piano()
else:
    instr = Instrument()

instr.play()
```

**Q-16  Create a class Animal with an abstract method sound(). Create two derived classes Cow and Sheep that implement the sound() method.**

```python
print("\n16. Abstract Animal Sound")

from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Cow(Animal):
    def sound(self):
        print("Cow says Moo!")

class Sheep(Animal):
    def sound(self):
        print("Sheep says Baa!")

animal_type = input("Enter animal (cow/sheep): ").strip().lower()

if animal_type == "cow":
    a = Cow()
elif animal_type == "sheep":
    a = Sheep()
else:
    print("Unknown animal")
    a = None

if a:
    a.sound()
```

**Q-18 Create a class House with attributes address and price. Create a derived class Mansion that adds an attribute number_of_rooms.**

```python
print("\n18. House and Mansion")

class House:
    def __init__(self, address, price):
        self.address = address
        self.price = price

    def show_info(self):
        print(f"House at {self.address} costs ${self.price}")

class Mansion(House):
    def __init__(self, address, price, number_of_rooms):
        super().__init__(address, price)
        self.number_of_rooms = number_of_rooms

    def show_info(self):
        print(f"Mansion at {self.address} costs ${self.price} with {self.number_of_rooms} rooms.")

# User input
address = input("Enter the address: ")
price = float(input("Enter the price: "))
rooms = int(input("Enter number of rooms: "))

# Create a Mansion object
m = Mansion(address, price, rooms)

# Show mansion info
m.show_info()
```
